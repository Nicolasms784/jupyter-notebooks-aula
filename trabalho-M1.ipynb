### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

dx = abs(x2 - x1)
dy = abs(y2 - y1)

distancia = math.sqrt(dx ** 2 + dy ** 2)
area = math.pi * distancia ** 2

print()
print("dx =", round(dx, 2))
print("dy =", round(dy, 2))
print("Distância entre os pontos:", round(distancia, 2))
print("Área do círculo de raio", round(distancia, 2), ":", round(area, 2))
````

````python
from jupyturtle import *


def poligono(lados, tamanho):
    """Desenha um polígono regular com a quantidade de lados e tamanho informados.

    lados: número de lados do polígono.
    tamanho: comprimento de cada lado.
    """
    angulo = 360 / lados

    for i in range(lados):
        forward(tamanho)
        left(angulo)


def flor(petalas, lados, tamanho):
    """Desenha uma flor reutilizando a função poligono.

    petalas: quantidade de pétalas da flor.
    lados: quantidade de lados de cada pétala.
    tamanho: comprimento dos lados das pétalas.
    """
    angulo = 360 / petalas

    for i in range(petalas):
        poligono(lados, tamanho)
        left(angulo)
````

**(b)** Escreva uma função chamada `eh_triangulo(a, b, c)` que retorne `True` se três medidas podem formar um
triângulo. A condição é que cada lado seja **positivo** e que cada lado seja **menor** que a soma dos outros dois.

**(c)** Usando `for i in range(...)`, exiba para os anos de **1996 a 2005** se cada um é bissexto ou não.

...
eh_triangulo(3, 4, 5) -> True

````python
def eh_bissexto(ano):
    """Retorna True se o ano for bissexto e False caso contrário."""
    return (ano % 400 == 0) or ((ano % 4 == 0) and not (ano % 100 == 0))


def eh_triangulo(a, b, c):
    """Retorna True se os três lados puderem formar um triângulo."""
    return (a > 0 and b > 0 and c > 0) and (a < b + c and b < a + c and c < a + b)


for i in range(1996, 2006):
    print(i, "é bissexto?", eh_bissexto(i))


print()
print("eh_triangulo(3, 4, 5) ->", eh_triangulo(3, 4, 5))
print("eh_triangulo(1, 2, 10) ->", eh_triangulo(1, 2, 10))
print("eh_triangulo(-1, 4, 5) ->", eh_triangulo(-1, 4, 5))
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
def classificar_imc(peso, altura):
    """Calcula o IMC e retorna o IMC junto com sua classificação."""
    if peso <= 0 or altura <= 0:
        return None

    imc = peso /(altura ** 2)

    if imc < 18.5:
        classificacao = "Abaixo do peso"
    elif imc < 25.0:
        classificacao = "Peso normal"
    elif imc < 30.0:
        classificacao = "Sobrepeso"
    else:
        classificacao = "Obesidade"

    return imc, classificacao


peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))

resultado = classificar_imc(peso, altura)

if resultado is None:
    print("Dados inválidos: peso e altura devem ser positivos.")
else:
    imc, classificacao = resultado
    print("IMC:", format(imc, ".2f"), "->", classificacao)
````

````python
def media_notas(nota1, nota2, nota3):
    """Calcula e retorna a média de três notas."""
    return (nota1 + nota2 + nota3) / 3


def situacao(media):
    """Retorna a situação do aluno de acordo com sua média."""
    if media >= 7:
        return "Aprovado"
    elif media >= 4:
        return "Recuperação"
    else:
        return "Reprovado"


quantidade_alunos = int(input("Quantos alunos? "))

nomes = []
medias = []
situacoes = []

for i in range(quantidade_alunos):
    print()
    print("Aluno", i + 1, "-")

    nome = input("Nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))

    media = media_notas(nota1, nota2, nota3)
    resultado = situacao(media)

    nomes.append(nome)
    medias.append(media)
    situacoes.append(resultado)


print()
print("===== BOLETIM =====")

for i in range(quantidade_alunos):
    print(nomes[i] + ":", format(medias[i], ".1f"), "-", situacoes[i])

print("===================")

media_geral = sum(medias) / quantidade_alunos

quantidade_aprovados = 0

for resultado in situacoes:
    if resultado == "Aprovado":
        quantidade_aprovados += 1

percentual_aprovados = (quantidade_aprovados / quantidade_alunos) * 100

indice_maior_media = 0

for i in range(1, quantidade_alunos):
    if medias[i] > medias[indice_maior_media]:
        indice_maior_media = i

print("Média geral da turma:", format(media_geral, ".2f"))
print("Aprovados:", quantidade_aprovados, "de", quantidade_alunos,
      "(", format(percentual_aprovados, ".2f") + "%)")
print("Melhor média:", nomes[indice_maior_media],
      "(", format(medias[indice_maior_media], ".2f") + ")")
````
